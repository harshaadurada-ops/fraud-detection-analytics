```python

import pandas as pd
import os

file_path = "/content/fraud_dataset/creditcard.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python

print(df.columns.tolist())
print("\nMissing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python

print("Dataset shape:", df.shape)

print("\nMissing values:", df.isnull().sum().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nFraud class counts:")
print(df["Class"].value_counts())
```

```python

df_clean = df.drop_duplicates().copy()

print("Original records:", len(df))
print("Records after cleaning:", len(df_clean))
print("Duplicate records removed:", len(df) - len(df_clean))
```

```python

total_transactions = len(df_clean)

legitimate = int((df_clean["Class"] == 0).sum())
fraudulent = int((df_clean["Class"] == 1).sum())

legitimate_pct = legitimate / total_transactions * 100
fraud_pct = fraudulent / total_transactions * 100

print("Total transactions:", total_transactions)
print("Legitimate transactions:", legitimate)
print("Fraudulent transactions:", fraudulent)
print("Legitimate percentage:", round(legitimate_pct, 3), "%")
print("Fraud percentage:", round(fraud_pct, 3), "%")
```

```python

import matplotlib.pyplot as plt

counts = df_clean["Class"].value_counts().reindex([0, 1], fill_value=0)

plt.figure(figsize=(7, 5))
plt.bar(["Legitimate", "Fraudulent"], counts.values)

plt.title("Fraud vs Legitimate Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df_clean["Amount"], bins=50)

plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

df_clean.boxplot(column="Amount", by="Class")

plt.title("Transaction Amount by Fraud Class")
plt.suptitle("")
plt.xlabel("Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Transaction Amount")
plt.tight_layout()
plt.show()
```

```python

fraud_df = df_clean[df_clean["Class"] == 1]

plt.figure(figsize=(9, 5))
plt.hist(fraud_df["Time"], bins=40)

plt.title("Fraudulent Transactions Over Time")
plt.xlabel("Elapsed Time (seconds)")
plt.ylabel("Number of Fraudulent Transactions")
plt.tight_layout()
plt.show()
```

```python

average_amount = df_clean.groupby("Class")["Amount"].mean()

print(average_amount)

average_amount.plot(kind="bar", rot=0, figsize=(7, 5))
plt.title("Average Transaction Amount by Class")
plt.xlabel("Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Average Amount")
plt.xticks([0, 1], ["Legitimate", "Fraudulent"])
plt.tight_layout()
plt.show()
```

```python

X = df_clean.drop(columns=["Class"])
y = df_clean["Class"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Fraud cases in training:", int(y_train.sum()))
print("Fraud cases in testing:", int(y_test.sum()))
```

```python

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

model = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    )
)

model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    classification_report
)

y_pred = model.predict(X_test)
fraud_scores = model.predict_proba(X_test)[:, 1]

print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, y_pred, zero_division=0), 4))
print("Average precision:", round(average_precision_score(y_test, fraud_scores), 4))

print("\nDetailed classification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python

from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Legitimate", "Fraud"],
    yticklabels=["Legitimate", "Fraud"]
)

plt.title("Fraud Detection Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.show()

print("Confusion Matrix:")
print(cm)
```

```python

results = X_test.copy()

results["Actual_Class"] = y_test.to_numpy()
results["Predicted_Class"] = y_pred
results["Fraud_Score"] = fraud_scores

results["Actual_Status"] = results["Actual_Class"].map({
    0: "Legitimate",
    1: "Fraud"
})

results["Predicted_Status"] = results["Predicted_Class"].map({
    0: "Legitimate",
    1: "Fraud"
})

display(results.head())

results.to_csv(
    "/content/fraud_detection_results.csv",
    index=False
)

print("Prediction results saved successfully!")
```
